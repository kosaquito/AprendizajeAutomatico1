#DATASETS DE ANALISIS DE SENTIMIENTO

```python
## prompt: Descargar el .zip desde https://archive.ics.uci.edu/static/public/331/sentiment+labelled+sentences.zip, descomprimirlo, cargar en variable dataset el archivo csv
import pandas as pd
import os

# Descargar el archivo zip
!wget https://archive.ics.uci.edu/static/public/331/sentiment+labelled+sentences.zip

# Descomprimir el archivo zip
!unzip sentiment+labelled+sentences.zip

# Listar los archivos para verificar
!ls

# Define the directory where files were unzipped
extracted_dir = 'sentiment labelled sentences'

# Lista de archivos CSV potenciales en el extraido
# Use os.path.join to create the full path
csv_files = [os.path.join(extracted_dir, file) for file in os.listdir(extracted_dir) if file.endswith('.txt') and 'amazon_cells' in file]

if csv_files:
    # Assuming the first found file is the correct one (amazon_cells_labelled.txt)
    file_to_load = csv_files[0]
    print(f"Loading file: {file_to_load}")
    # These datasets are often tab-separated without a header
    dataset = pd.read_csv(file_to_load, sep='\t', header=None, names=['text', 'sentiment'])
    print("Dataset loaded successfully:")
    print(dataset.head())
else:
    print("No suitable .txt file found after unzipping. Please check the extracted files.")
```
